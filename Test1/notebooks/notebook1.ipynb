```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data
df = pd.read_csv('sales_data.csv', parse_dates=['Date'])

# Display first few rows
df.head()
```

Total Sales by Region

```python
region_sales = df.groupby('Region')['Sales'].sum().reset_index()

sns.barplot(data=region_sales, x='Region', y='Sales')
plt.title("Total Sales by Region")
plt.show()
```

Sales Trend Over Time

```python
date_sales = df.groupby('Date')['Sales'].sum().reset_index()

plt.plot(date_sales['Date'], date_sales['Sales'], marker='o')
plt.title("Sales Over Time")
plt.xlabel("Date")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()
```

Sales by Product and Region

```python
pivot = df.pivot_table(index='Region', columns='Product', values='Sales', aggfunc='sum')

pivot.plot(kind='bar', stacked=True)
plt.title("Sales by Product per Region")
plt.ylabel("Sales")
plt.show()
```
